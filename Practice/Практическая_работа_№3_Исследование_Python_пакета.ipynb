Вы будете работать с пакетом `calculator` из раздела №16:

```

    ├── basic/
    │   ├── __init__.py
```

Создайте эти файлы в папке проекта вручную или запустите в Colab ячейку ниже: она создаст пакет и скачает архив `calculator_project.zip`.

````python
# @title Создать пакет calculator и скачать архив
import shutil
from pathlib import Path

FILES = {
    "calculator/__init__.py": '__all__ = ["basic", "advanced"]\n',
    "calculator/basic/__init__.py": 'from .addition import add\nfrom .subtraction import subtract\n',
    "calculator/basic/addition.py": 'def add(a, b):\n    return a + b\n',
    "calculator/basic/subtraction.py": 'def subtract(a, b):\n    return a - b\n',
    "calculator/advanced/__init__.py": 'from .exponentiation import power\nfrom .root import square_root\n',
    "calculator/advanced/exponentiation.py": 'def power(a, b):\n    return a ** b\n',
    "calculator/advanced/root.py": 'def square_root(a):\n    return a ** 0.5\n',
    "main.py": 'from calculator.basic import add, subtract\nfrom calculator.advanced import power, square_root\n\nprint(add(2, 3))          # 5\nprint(subtract(5, 2))     # 3\nprint(power(2, 3))        # 8\nprint(square_root(16))    # 4.0\n',
}

root = Path("calculator_project")
for name, text in FILES.items():
    path = root / name
    path.parent.mkdir(parents=True, exist_ok=True)
    path.write_text(text, encoding="utf-8")

shutil.make_archive("calculator_project", "zip", root_dir=".", base_dir="calculator_project")
from google.colab import files
files.download("calculator_project.zip")
````

<font color='red'>Я заменил относительные импорты на абсолютные и запустил main.py: результаты не изменились.
Относительный импорт (from .addition import add) указывает путь от текущего модуля: точка означает «текущий пакет».
Плюсы: короче; при переименовании или переносе всего пакета в другое место импорты внутри него не ломаются, потому что они не содержат имя пакета.
Минусы: путь менее наглядный, не сразу видно, откуда берётся имя; такие импорты работают только внутри пакета и не работают, если запустить файл напрямую как скрипт.
Абсолютный импорт (from calculator.basic.addition import add) указывает полный путь от корня пакета.
Плюсы: сразу понятно, откуда импортируется имя.
Минусы: длиннее; при переименовании пакета или папки нужно менять все такие импорты.</font>

<font color='red'>После переименования basic в simple придётся исправить абсолютные импорты, потому что в них прописано имя папки (calculator.basic... → calculator.simple...).
